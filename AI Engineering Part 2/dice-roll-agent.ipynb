#### Dice Roll Agent - Preliminary Dice Roll Agent

```python
#creating dice roll agent:
import random
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()

def dice_roll():
    return random.randint(1,6)

OPENAI_API_KEY = os.getenv(key='OPENAI_API_KEY')

client = OpenAI(api_key= OPENAI_API_KEY)
```

```python
# Get target number and max iterations
target = 6
max_iterations = 20
# Initialize history and iteration counter
history = ""
iteration = 0

while iteration < max_iterations:

    # Increment iteration
    iteration +=1

    # Build prompt using:
    instructions = f""" 
    You are playing a dice game. You need to roll a {target} on the dice.

    You've made the following rolls so far:
    {history if history else "No rolls yet."}

    Based on what you have rolled so far, decide your next action:
    Option A: if you have achieved the target, respond with exactly DONE
    Option B: if you have not acheived the target, respond with exactly: ROLL THE DICE
    
    """
  
    # Send prompt to the LLM
    response = client.chat.completions.create(
        model = 'gpt-4.1-mini',
        messages = [{'role':'user', 'content' : instructions}]
    )
    # Extract the model's response
    result = response.choices[0].message.content
    # print(f' this is {result} from llm')

    if "DONE" in result:
        print (result)
        break

    if "ROLL THE DICE" in result:
        roll = dice_roll()
        print(f"Rolled {roll}\n")
        # append roll to history
        history += f"Roll {iteration}: roll is {roll}\n"

# Print total iterations
print(f"GAME OVER: Rolled target in {iteration - 1}")
```
